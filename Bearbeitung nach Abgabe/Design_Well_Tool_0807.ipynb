```python
def calc_Q_vol(P_gw, delta_T):
    """Berechnung des benötigten Fördervolumenstroms in m³/h.
    Args:
        P_gw (float): Entzugsleistung des Grundwassers in W
        delta_T (float): Temperaturdifferenz in K"""
    rho = 1000  # Dichte in kg/m³
    c = 4180  # Spezifische Wärmekapazität in J/kgK
    Q_vol = (P_gw) / (rho * c * delta_T) #W/(J/m3)--> m3/s
    Q_vol_m3h = Q_vol * 3600  # Umrechnung von m³/s auf m³/h
    Q_vol_Ls = Q_vol * 1000 # Umrechnung von m³/s auf L/s
    return round(Q_vol, 5), round(Q_vol_m3h, 5), round(Q_vol_Ls, 5)

Q_vol_h_value = calc_Q_vol(P_gw_h_max, T_slider.value)
if "Kühlen" in betrieb.value:
    Q_vol_k_value = calc_Q_vol(P_gw_k_max, T_slider.value)

output_GW_vol = widgets.Output()

if "Kühlen" in betrieb.value:
    with output_GW_vol:
        if "Kühlen" in betrieb.value:
            max_value = round(max(Q_vol_h_value[1], Q_vol_k_value[1]), 2)
        else:
            max_value = round(Q_vol_h_value[1], 2)
        display(HTML(f"""
        <h3 style='color:cornflowerblue'>Berechnung der erforderlichen Grundwasser-Förderrate</h3>
        <div style="
            background:whitesmoke;
            padding:15px;
            border-radius:12px;
            width:450px;
            font-size:14px;
            line-height:1.6">
            
            Die Grundwasser-Förderrate für den Heizbetrieb:</>
            <div style="color:black">
                {Q_vol_h_value[0]} m³/s<br>
                {Q_vol_h_value[1]} m³/h<br>
                {Q_vol_h_value[2]} L/s
            </div>
            <br>
            Die Grundwasser-Förderrate für den Kühlbetrieb:</>
            <div style="color:black">
                {Q_vol_k_value[0]} m³/s<br>
                {Q_vol_k_value[1]} m³/h<br>
                {Q_vol_k_value[2]} L/s
            </div>  
            <br>
            <b>Die maximal erforderliche GW-Förderrate beträgt:</b>
            <span style="color:black">
               <b>{round(max_value,2)}</b> <b>m³/h</b>
            </span>
        </div>
        """))

else:
    with output_GW_vol:
        display(HTML(f"""
        <h3 style='color:cornflowerblue'>Berechnung der erforderlichen Grundwasser-Förderrate</h3>
        <div style="
            background:whitesmoke;
            padding:15px;
            border-radius:12px;
            width:450px;
            font-size:14px;
            line-height:1.8">

            Die Grundwasser-Förderrate für den Heizbetrieb:<br>

            <div style="color:black">
                {Q_vol_h_value[0]} m³/s<br>
                {Q_vol_h_value[1]} m³/h<br>
                {Q_vol_h_value[2]} L/s
            </div>
            <br>
            <b>Die maximal erforderliche GW-Förderrate beträgt:</b>
            <span style="color:black">
                <b>{round(Q_vol_h_value[1],2)}</b> <b>m³/h</b>
            </span>

        </div>
        """))

# display(output_GW_vol)
```

```python
#TEST
def calc_Q_vol_range(P_gw, delta_T, unsicherheit=0.1):
    """Berechnet Volumenstrombereich (m³/s, m³/h, L/s) mit Unsicherheit ±%"""

    rho = 1000
    c = 4180

    factor_min = 1 - unsicherheit
    factor_max = 1 + unsicherheit

    def calc(P):
        Q_vol = P / (rho * c * delta_T)  # m³/s
        return Q_vol, Q_vol * 3600, Q_vol * 1000

    Q_min = calc(P_gw * factor_min)
    Q_max = calc(P_gw * factor_max)

    return {
        "min": Q_min,
        "max": Q_max
    }
Q_vol_h = calc_Q_vol_range(P_gw_h_max, T_slider.value)

if "Kühlen" in betrieb.value:
    Q_vol_k = calc_Q_vol_range(P_gw_k_max, T_slider.value)
output_GW_vol = widgets.Output()

with output_GW_vol:

    if "Kühlen" in betrieb.value:
        Q_max = max(Q_vol_h["max"][1], Q_vol_k["max"][1])
        Q_min = min(Q_vol_h["min"][1], Q_vol_k["min"][1])

        display(HTML(f"""
        <h3 style='color:cornflowerblue'>Berechnung der erforderlichen Grundwasser-Förderrate</h3>
        <div style="
            background:whitesmoke;
            padding:15px;
            border-radius:12px;
            width:450px;
            font-size:15px;
            line-height:1.8">

            <b>Für den Heizbetrieb:</b>
            {round(Q_vol_h["min"][1],2)} – {round(Q_vol_h["max"][1],2)} m³/h
            <br>
            <b>Für den Kühlbetrieb:</b>
            {round(Q_vol_k["min"][1],2)} – {round(Q_vol_k["max"][1],2)} m³/h

            <hr>

            <b>Maximaler Auslegungsbereich:</b><br>
            {round(Q_min,2)} – {round(Q_max,2)} m³/h

        </div>
        """))

    else:
        display(HTML(f"""
        <h3 style='color:cornflowerblue'>Berechnung der erforderlichen Grundwasser-Förderrate</h3>
        <div style="
            background:whitesmoke;
            padding:15px;
            border-radius:12px;
            width:450px;
            font-size:15px;
            line-height:1.8">

            <b>Für den Heizbetrieb:</b>
            {round(Q_vol_h["min"][1],2)} – {round(Q_vol_h["max"][1],2)} m³/h

        </div>
        """))

display(output_GW_vol)
```

## 2.2. Fassungsvermögen Qf und Wasserandrang Qa

```python
# --- Berechnung Qa, Qf und Fehler ---

# Variablen & kleine Rechnungen
# Absenkung Datenreihe zur Visualisierung
s_array = np.arange(0.1, 0.9*hf_wert.value, 0.05) # in m
# Berechnung des effektiven Brunnenradius
calc_r_e=round((r_bl_wert.value + r_br_wert.value)/2, 3) #in m
# Symbole definieren
kf, hF, s_sym, r_eff, h0 = sp.symbols('kf hF s r_eff h0')
# Berechnung der Absenkungsreichweite
R = 3000 * s_sym * sp.sqrt(kf)
# Berechnung der maximalen Brunnneneintrittsgeschwindigkeit
v_max = math.sqrt(kf_wert.value) / 15

# Definition Fehler
delta_hF = hf_fehler.value
delta_h0 = h0_fehler.value

# --- Ungespannter Grundwasserleiter ---
Q_a_expr_ug = sp.pi * kf * ((hF**2 - (hF - s_sym)**2) / sp.log(R / r_eff))
# Ableitungen
dQa_dkf_expr_ug = sp.diff(Q_a_expr_ug, kf)
dQa_dhF_expr_ug = sp.diff(Q_a_expr_ug, hF)
Qa_func_ug = sp.lambdify((s_sym, kf, hF, r_eff), Q_a_expr_ug, "numpy")
dQa_dkf_func_ug = sp.lambdify((s_sym, kf, hF, r_eff), dQa_dkf_expr_ug, "numpy")
dQa_dhF_func_ug = sp.lambdify((s_sym, kf, hF, r_eff), dQa_dhF_expr_ug, "numpy")

def calc_delta_Qa_ug(s_val, kf_val, hF_val, r_eff_val, log_width):
    """calc_delta_Qa_ug berechnet Wasserandrang Qa mit Fehler für ungespannte GWL."""
    Q_a_ug = Qa_func_ug(s_val, kf_val, hF_val, r_eff_val)
    dQa_dkf_val_ug = dQa_dkf_func_ug(s_val, kf_val, hF_val, r_eff_val)
    dQa_dhF_val_ug = dQa_dhF_func_ug(s_val, kf_val, hF_val, r_eff_val)
    # log-Raum
    sigma_log10_ug = log_width / math.sqrt(12)
    # Ableitung nach log10(kf)
    dQa_dlogkf_ug = dQa_dkf_val_ug * kf_val * math.log(10)
    delta_Qa_ug = math.sqrt((dQa_dlogkf_ug * sigma_log10_ug)**2 + (dQa_dhF_val_ug * delta_hF)**2)
    return float(Q_a_ug), float(delta_Qa_ug)

def calc_delta_Qf_ug(s_val, kf_val, hF_val, r_bl_val, log_width):
    """calc_delta_Qf_ug berechnet das Fassungsvermögen Qf mit Fehler für ungespannte GWL."""
    # Formel für Qf
    Q_f_ug = 2 * r_bl_val * math.pi * (hF_val - s_val) * v_max
    dQf_dkf_ug = math.pi * r_bl_val * (hF_val - s_val) / (15 * math.sqrt(kf_val))
    dQf_dhF_ug = 2 * math.pi * r_bl_val * math.sqrt(kf_val) / 15
    # log-Raum
    sigma_log10_ug = log_width / math.sqrt(12)
    # Ableitung nach log10(kf)
    dQf_dlogkf_ug = dQf_dkf_ug * kf_val * math.log(10)
    delta_Qf_ug = math.sqrt((dQf_dlogkf_ug * sigma_log10_ug)**2 +(dQf_dhF_ug * delta_hF)**2)
    return float(Q_f_ug), float(delta_Qf_ug)

# --- Gespannter Grundwasserleiter ---
Q_a_expr_g = 2 * sp.pi * hF * kf * ((h0 - (h0 - s_sym)) / sp.log(R / r_eff))

# Ableitungen
dQ_dkf_expr_g = sp.diff(Q_a_expr_g, kf)
dQ_dhF_expr_g = sp.diff(Q_a_expr_g, hF)

Qa_func_g = sp.lambdify((s_sym, kf, hF, r_eff, h0), Q_a_expr_g, "numpy")
dQ_dkf_func_g = sp.lambdify((s_sym, kf, hF, r_eff, h0), dQ_dkf_expr_g, "numpy")
dQ_dhF_func_g = sp.lambdify((s_sym, kf, hF, r_eff, h0), dQ_dhF_expr_g, "numpy")

def calc_delta_Qa_g(s_val, kf_val, hF_val, r_eff_val, h0_val, log_width):
    """calc_delta_Qa_g berechnet Wasserandrang Qa mit Fehler für gespannte GWL."""
    Q_a_g = Qa_func_g(s_val, kf_val, hF_val, r_eff_val, h0_val)
    dQ_dkf_val_g = dQ_dkf_func_g(s_val, kf_val, hF_val, r_eff_val, h0_val)
    dQ_dhF_val_g = dQ_dhF_func_g(s_val, kf_val, hF_val, r_eff_val, h0_val)
    # log-Raum
    sigma_log10_g = log_width / math.sqrt(12)
    # Ableitung nach log10(kf)
    dQ_dlogkf_g = dQ_dkf_val_g * kf_val * math.log(10)
    delta_Qa_g = math.sqrt((dQ_dlogkf_g * sigma_log10_g)**2 + (dQ_dhF_val_g * delta_hF)**2)
    return float(Q_a_g), float(delta_Qa_g)

def calc_delta_Qf_g(s_val, kf_val, hF_val, r_bl_val, log_width):
    """calc_delta_Qf_g berechnet das Fassungsvermögen Qf mit Fehler für gespannte GWL."""
    # Formel für Qf
    Q_f_g = 2 * r_bl_val * math.pi * (hF_val - s_val) * v_max
    dQ_dkf_g = math.pi * r_bl_val * (hF_val - s_val) / (15 * math.sqrt(kf_val))
    dQ_dhF_g = 2 * math.pi * r_bl_val * math.sqrt(kf_val) / 15
    # log-Raum
    sigma_log10_g = log_width / math.sqrt(12)
    # Ableitung nach log10(kf)
    dQ_dlogkf_g = dQ_dkf_g * kf_val * math.log(10)
    delta_Qf_g = math.sqrt((dQ_dlogkf_g * sigma_log10_g)**2 +(dQ_dhF_g * delta_hF)**2)
    return float(Q_f_g), float(delta_Qf_g)
```

```python
# Berechnung Qf, Qa und Unsicherheiten für s_array, in Dataframe df_Q (UNGESPANNTER GWL)
df_Q_ug = pd.DataFrame({'s [m]': s_array})
df_Q_ug[['Q_a [m3/s]', 'ΔQ_a [m3/s]']] = df_Q_ug['s [m]'] \
    .apply(lambda s: calc_delta_Qa_ug(s, kf_wert.value, hf_wert.value, calc_r_e, kf_fehler.value)) \
    .apply(pd.Series)

df_Q_ug[['Q_f [m3/s]', 'ΔQ_f [m3/s]']] = df_Q_ug['s [m]'] \
    .apply(lambda s: calc_delta_Qf_ug(s, kf_wert.value, hf_wert.value, r_bl_wert.value, kf_fehler.value)) \
    .apply(pd.Series)

df_Q_ug = df_Q_ug[['s [m]', 'Q_f [m3/s]', 'ΔQ_f [m3/s]', 'Q_a [m3/s]', 'ΔQ_a [m3/s]']]

# Berechnung Qf, Qa und Unsicherheiten für s_array, in Dataframe df_Q_g (GESPANNTER GWL)
df_Q_g = pd.DataFrame({'s [m]': s_array})
df_Q_g[['Q_a [m3/s]', 'ΔQ_a [m3/s]']] = df_Q_g['s [m]'] \
    .apply(lambda s: calc_delta_Qa_g(s, kf_wert.value, hf_wert.value, calc_r_e, h0_wert.value, kf_fehler.value)) \
    .apply(pd.Series)

df_Q_g[['Q_f [m3/s]', 'ΔQ_f [m3/s]']] = df_Q_g['s [m]'] \
    .apply(lambda s: calc_delta_Qf_g(s, kf_wert.value, hf_wert.value, r_bl_wert.value, kf_fehler.value)) \
    .apply(pd.Series)

df_Q_g = df_Q_g[['s [m]', 'Q_f [m3/s]', 'ΔQ_f [m3/s]', 'Q_a [m3/s]', 'ΔQ_a [m3/s]']]

if type_gwl == 'Ungespannt':
    print("Ungespannter Grundwasserleiter:")
    display(df_Q_ug.head())
else:
    print("Gespannter Grundwasserleiter:")
    display(df_Q_g.head())
```

## 2.3. Optimale Brunnenleistung Qopt

```python
Q_erford_min = calc_Q_vol_range(P_gw_h_max, T_slider.value, unsicherheit=0.1)["min"][1] 
Q_erford_max = calc_Q_vol_range(P_gw_h_max, T_slider.value, unsicherheit=0.1)["max"][1]
Q_mittel = (Q_erford_min + Q_erford_max) / 2
```

```python
if type_gwl == 'Ungespannt':
    df_Q = df_Q_ug
else:
    df_Q = df_Q_g

x = np.array(df_Q['s [m]'])

y_f = np.array(df_Q['Q_f [m3/s]'])
dy_f = np.array(df_Q['ΔQ_f [m3/s]'])

y_a = np.array(df_Q['Q_a [m3/s]'])
dy_a = np.array(df_Q['ΔQ_a [m3/s]'])

factor = 3600  # Umrechnungsfaktor m³/s -> m³/h

f1 = interp1d(x, y_f, kind='linear')
f2 = interp1d(x, y_a, kind='linear')

def diff(s):
    return f1(s) - f2(s)

xs = np.linspace(x.min(), x.max(), 2000)
vals = diff(xs)

idx = np.where(np.sign(vals[:-1]) != np.sign(vals[1:]))[0]

schnittpunkt_x = []
schnittpunkt_y = []

for i in idx:
    try:
        s_opt = brentq(diff, xs[i], xs[i+1])
        Q_opt = float(f1(s_opt))
        schnittpunkt_x.append(s_opt)
        schnittpunkt_y.append(Q_opt)
    except ValueError:
        pass

if len(schnittpunkt_x) == 0:
    raise ValueError("Kein Schnittpunkt gefunden")

s_opt = float(schnittpunkt_x[0])
Q_opt = float(schnittpunkt_y[0])

f_f = interp1d(x, y_f, kind='linear')
f_a = interp1d(x, y_a, kind='linear')
f_df = interp1d(x, dy_f, kind='linear')
f_da = interp1d(x, dy_a, kind='linear')

Qf = f_f(xs)
Qa = f_a(xs)

dQf = f_df(xs)
dQa = f_da(xs)

Qf_low = Qf - dQf
Qf_high = Qf + dQf

Qa_low = Qa - dQa
Qa_high = Qa + dQa

mask = (Qf_low <= Qa_high) & (Qf_high >= Qa_low)

xs_overlap = xs[mask]

if len(xs_overlap) == 0:
    raise ValueError("Kein Überlappungsbereich gefunden")

s_left = xs_overlap[0]
s_right = xs_overlap[-1]

ds_left = s_opt - s_left
ds_right = s_right - s_opt

Q_low = np.maximum(Qf_low[mask], Qa_low[mask])
Q_high = np.minimum(Qf_high[mask], Qa_high[mask])

Q_low_min = np.min(Q_low)
Q_high_max = np.max(Q_high)

Q_opt_plot = Q_opt * factor

dQ_down = Q_opt_plot - Q_low_min * factor
dQ_up = Q_high_max * factor - Q_opt_plot


fig, ax = plt.subplots(figsize=(8, 6))

# Plot Qf und Qa mit Unsicherheitsbereichen
ax.fill_between(x, (y_f - dy_f) * factor, (y_f + dy_f) * factor, color='royalblue', alpha=0.1)
ax.plot(x, y_f * factor, color='royalblue', label='Q_f')
ax.fill_between(x, (y_a - dy_a) * factor, (y_a + dy_a) * factor, color='springgreen', alpha=0.1)
ax.plot(x, y_a * factor, color='springgreen', label='Q_a')

# Markierung der erforderlichen GW-Förderrate
ax.axhline(Q_mittel, color="darkorange", linestyle="--", label="Erforderliche GW-Förderrate")
ax.fill_between(x, Q_vol_h["min"][1], Q_vol_h["max"][1], color="darkorange", alpha=0.2)

# Optimaler Betriebspunkt mit Fehlerbalken
ax.errorbar(s_opt, Q_opt_plot, xerr=[[ds_left], [ds_right]], yerr=[[dQ_down], [dQ_up]],
    fmt='o', color='mediumvioletred', ecolor='mediumvioletred', elinewidth=2, capsize=6,
    label='Optimaler Betriebspunkt')

ax.annotate(
    f"Q_opt = {Q_opt_plot:.2f} (+{dQ_up:.2f}/-{dQ_down:.2f}) m³/h\n"
    f"s_opt = {s_opt:.2f} (+{ds_right:.2f}/-{ds_left:.2f}) m",
    (s_opt, Q_opt_plot),
    xytext=(20, 20), textcoords="offset points",
    bbox=dict(boxstyle="round", fc="white", ec="mediumvioletred", alpha=0.9), color="mediumvioletred")

ax.set_xlabel("Absenkung s [m]")
ax.set_ylabel("Q [m³/h]")
ax.set_title("Betriebspunkt mit Unsicherheitsbereich")
ax.set_ylim(0, 15)
ax.set_xlim(0, 6)
ax.grid()
ax.legend(loc='upper left')

plt.show()
```

## 3.2. Rückströmung

```python
def Q_d_unsafe(Q, b, d, k_f, i):
    """Berechnung der dimensionslosen Förderrate ß für angegebenen möglichen Abstand."""
    a = (d/2) 
    beta = round(Q / (np.pi * b * a * k_f * i),1)
    return beta
def recirc_Q(beta):
    """recirc_Q berechnet das Verhältnis des vom Injektionsbrunnen zum Förderbrunnen zurück geströmten Grundwassers zu dem ungestörten Grundwasser.
    Args:
        beta (float): Dimensionslose Förderrate nach gewähltem Abstand
    Returns:
        float: Anteil der Rückstromrate I/Q"""
    I_Q = (2/np.pi) * (-(1/beta)*np.sqrt(beta-1) + np.arctan(np.sqrt(beta-1)))
    return I_Q
```
